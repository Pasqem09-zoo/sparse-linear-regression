```python
# vediamo questa matrice di cov con le correlazioni come esce
p=10
A = np.random.randn(p, p) #values from standard normal distribution
C = A @ A.T   # positive semidefinite matrix
print(C)
# correlation matrix (Corr)
diag_C = np.sqrt(np.diag(C))
Corr = C / np.outer(diag_C, diag_C)

# check per il numero di valori maggiori di 0.90 e minori di 0.10
print(np.sum(Corr > 0.90))
print(np.sum(Corr < -0.70))
```

```python
#test miqp
import numpy as np
from src.least_squares import LeastSquaresProblem
from src.miqp import MIQPSolver

# Dimensioni piccole
n = 20
p = 5
k = 2

np.random.seed(0)

# Genera matrice X
X = np.random.randn(n, p)

# Crea beta vero sparso
beta_true = np.zeros(p)
beta_true[0] = 3.0
beta_true[3] = -2.0

# Genera y senza rumore
y = X @ beta_true

# Crea problema
problem = LeastSquaresProblem(X, y)

# Risolvi con MIQP
solver = MIQPSolver(problem, k)
solver.solve()
beta_miqp = solver.get_solution()

print("Beta true:", beta_true)
print("Beta MIQP:", beta_miqp)
print("Errore:", np.linalg.norm(beta_miqp - beta_true))
print("Numero non zero trovati:", np.sum(np.abs(beta_miqp) > 1e-6))
```
